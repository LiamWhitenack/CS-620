# Noun chunks

```python
# !python -m spacy download en_core_web_lg
import nltk
import spacy
from nltk import word_tokenize
from nltk.corpus import stopwords
from string import punctuation
small_model = spacy.load("en_core_web_sm")
large_model = spacy.load("en_core_web_lg")
```

```python
def print_noun_chunks(text, model):
    doc = model(text)
    for noun_chunk in doc.noun_chunks:
        print(noun_chunk.text)
```

```python
def read_text_file(filename):
    file = open(filename, "r", encoding="utf-8")
    return file.read()
```

```python
sherlock_holmes_part_of_text = read_text_file("sherlock_holmes_1.txt")
print_noun_chunks(sherlock_holmes_part_of_text, small_model)
```

# Explore noun chunk properties

```python
def explore_properties(sentence, model):
    doc = model(sentence)
    other_span = "emotions"
    other_doc = model(other_span)
    for noun_chunk in doc.noun_chunks:
        print(noun_chunk.text)
        print("Noun chunk start and end", "\t", noun_chunk.start, "\t", noun_chunk.end)
        print("Noun chunk sentence:", noun_chunk.sent)
        print("Noun chunk root:", noun_chunk.root.text)
        print(f"Noun chunk similarity to '{other_span}'", noun_chunk.similarity(other_doc))
    print(f"Similarity of the sentence '{sentence}' to '{other_span}':", doc.similarity(other_doc))
```

```python
sentence = "All emotions, and that one particularly, were abhorrent to his cold, precise but admirably balanced mind."
explore_properties(sentence, small_model)
```

```python
sentence = "All emotions, and that one particularly, were abhorrent to his cold, precise but admirably balanced mind."
explore_properties(sentence, large_model)
```

Understanding the start and end indices of noun chunks is very useful in various natural language processing applications. Here's why:

*   **Information Extraction:** If you're trying to extract specific entities or concepts from text, the indices allow you to pinpoint the exact span of the noun chunk, making it easier to isolate and process the relevant information. For example, extracting all product names or person names from a document.
*   **Relationship Extraction:** Once you've identified noun chunks (which often represent entities), their positions in the sentence (indicated by indices) are crucial for identifying and extracting relationships between them. For example, finding sentences where a "person" noun chunk is followed by a "verb" and then a "location" noun chunk to identify where someone went.
*   **Text Summarization:** Noun chunks often represent key concepts. Identifying and extracting these chunks using their indices can be a step in generating summaries, either by selecting important sentences containing key noun chunks or by creating a summary based on the most frequent or significant noun chunks.
*   **Question Answering:** In question answering systems, identifying noun chunks in both the question and the source text can help match relevant parts of the text to answer the question. The indices help in extracting the precise answer span from the text.
*   **Sentiment Analysis:** Noun chunks often represent the targets of sentiment. By identifying the noun chunk and its associated sentiment (expressed by nearby adjectives or verbs), you can determine the sentiment towards specific entities. The indices help in linking the sentiment to the correct noun chunk.
*   **Coreference Resolution:** Identifying when different noun chunks refer to the same entity (e.g., "John," "he," "the man") often relies on analyzing their position and relationship to other parts of the text, where indices play a role.
*   **Text Generation:** When generating text, knowing the structure of noun chunks and their typical positions can help create more grammatically correct and meaningful sentences.

In essence, the indices provide the precise location of these meaningful units within the text, which is fundamental for many NLP tasks that require more than just token-level processing.
