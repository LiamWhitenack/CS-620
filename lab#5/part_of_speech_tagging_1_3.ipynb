# Part of speech tagging using spaCy

```python
def pos_tag_spacy(text, model):
    doc = model(text)
    words = [token.text for token in doc]
    pos = [token.pos_ for token in doc]
    return list(zip(words, pos))
```

```python
def read_text_file(filename):
    """Reads a text file and returns its content."""
    with open(filename, 'r') as f:
        return f.read()
```

```python
text = read_text_file("sherlock_holmes_1.txt")
words_with_pos = pos_tag_spacy(text, small_model)
print(words_with_pos)
```

```python
import nltk
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

nltk.download('averaged_perceptron_tagger')

```

```python
def word_tokenize_nltk(text):
    return nltk.tokenize.word_tokenize(text)

def word_tokenize_spacy(text, model):
    doc = model(text)
    return [token.text for token in doc]
```

```python
#nltk.download('averaged_perceptron_tagger') # Run the first time you run the notebook
def pos_tag_nltk(text):
    words = word_tokenize_nltk(text)
    words_with_pos = nltk.pos_tag(words)
    return words_with_pos
```

# Get part of speech tags using GPT-3

```python
from openai import OpenAI
client = OpenAI(api_key=OPEN_AI_KEY)
```

```python
prompt="""Decide what the part of speech tags are for a sentence.
Preserve original capitalization.
Return the list in the format of a python tuple: (word, part of speech).
Sentence: In his eyes she eclipses and predominates the whole of her sex."""
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    temperature=0,
    max_tokens=256,
    top_p=1.0,
    frequency_penalty=0,
    presence_penalty=0,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ],
)
print(response)
```

```python
print(response.choices[0].message.content)
```

```python
from ast import literal_eval

def pos_tag_gpt(text, client):
    prompt = f"""Decide what the part of speech tags are for a sentence.
    Preserve original capitalization.
    Return the list in the format of a python tuple: (word, part of speech).
    Do not include any other explanations.
    Sentence: {text}."""

    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        temperature=0,
        max_tokens=256,
        top_p=1.0,
        frequency_penalty=0,
        presence_penalty=0,
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
    )
    result = response.choices[0].message.content
    result = result.replace("\n", "")
    result = list(literal_eval(result))
    return result
```

```python
start = time.time()
first_sentence = "In his eyes she eclipses and predominates the whole of her sex."
words_with_pos = pos_tag_gpt(first_sentence, client)
print(words_with_pos)
print(f"GPT: {time.time() - start} s")
```

```python
words_with_pos_nltk = pos_tag_nltk(first_sentence)
print(words_with_pos_nltk)
```

This code snippet utilizes several powerful libraries for Natural Language Processing (NLP):

*   **`!python -m spacy download en_core_web_lg`**: This is a shell command that downloads the large English language model for spaCy. spaCy models include pre-trained pipelines for tasks like tokenization, part-of-speech tagging, and named entity recognition. The `en_core_web_lg` model is a larger model with more vocabulary and better accuracy compared to smaller models.

*   **`import nltk`**: Imports the Natural Language Toolkit (NLTK) library, a widely used platform for building Python programs to work with human language data.

*   **`import spacy`**: Imports the spaCy library, which is an industrial-strength library for NLP in Python. It's known for its speed and efficiency.

*   **`from nltk import word_tokenize`**: Imports the `word_tokenize` function specifically from NLTK's tokenize module. This function is used to split text into words.

*   **`from nltk.corpus import stopwords`**: Imports the `stopwords` corpus from NLTK. Stop words are common words (like 'the', 'a', 'is') that are often removed from text before processing because they don't usually carry much meaning.

*   **`from string import punctuation`**: Imports the `punctuation` string from Python's built-in `string` module. This string contains common punctuation characters.

*   **`small_model = spacy.load("en_core_web_sm")`**: Loads the small English language model (`en_core_web_sm`) from spaCy. This model is faster but less accurate than the large model.

*   **`large_model = spacy.load("en_core_web_lg")`**: Loads the large English language model (`en_core_web_lg`) from spaCy that was downloaded earlier.
