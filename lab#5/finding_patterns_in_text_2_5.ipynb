# Finding patterns

```python
# !python -m spacy download en_core_web_lg
import nltk
import spacy
from nltk import word_tokenize
from nltk.corpus import stopwords
from string import punctuation
small_model = spacy.load("en_core_web_sm")
large_model = spacy.load("en_core_web_lg")
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from spacy.matcher import Matcher
matcher = Matcher(small_model.vocab)
```

```python
patterns = [
    [{"POS": "VERB"}],
    [{"POS": "AUX"}, {"POS": "VERB"}],
    [{"POS": "AUX"}, {"POS": "ADJ"}],
    [{"POS": "AUX"}, {"POS": "VERB"}, {"POS": "ADP"}]
]
matcher.add("Verb", patterns)
```

```python
def read_text_file(filename):
    """Reads a text file and returns its content."""
    with open(filename, 'r') as f:
        return f.read()
```

```python
sherlock_holmes_part_of_text = read_text_file("sherlock_holmes_1.txt")
doc = small_model(sherlock_holmes_part_of_text)
```

```python
matches = matcher(doc)
for match_id, start, end in matches:
    string_id = small_model.vocab.strings[match_id]
    span = doc[start:end]
    print(match_id, string_id, start, end, span.text)
```

This code defines a list of patterns to be used with the spaCy `Matcher`. Each pattern is a list of dictionaries, where each dictionary specifies attributes a token must have to match that part of the pattern.

- `patterns`: This is a list containing four different patterns.
    - `[{"POS": "VERB"}]`: This pattern matches a single token with the part-of-speech tag "VERB" (a verb).
    - `[{"POS": "AUX"}, {"POS": "VERB"}]`: This pattern matches two consecutive tokens: the first must be an auxiliary verb ("AUX") and the second a verb ("VERB"). This could match phrases like "is running" or "has eaten".
    - `[{"POS": "AUX"}, {"POS": "ADJ"}]`: This pattern matches two consecutive tokens: the first an auxiliary verb ("AUX") and the second an adjective ("ADJ"). This could match phrases like "is happy" or "was sad".
    - `[{"POS": "AUX"}, {"POS": "VERB"}, {"POS": "ADP"}]`: This pattern matches three consecutive tokens: an auxiliary verb ("AUX"), followed by a verb ("VERB"), followed by an adposition ("ADP" - like a preposition or postposition). This could match phrases like "is looking at" or "was talking to".

- `matcher.add("Verb", patterns)`: This line adds the defined list of `patterns` to the `matcher` object. The string "Verb" is a unique ID assigned to this set of patterns. When the matcher is run on a document, it will find all occurrences of these patterns and report them with the ID "Verb".
