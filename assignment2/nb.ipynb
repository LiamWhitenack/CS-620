```python
from pandas import read_csv

actual = read_csv("data/actual.csv")
independent = read_csv("data/data_set_ALL_AML_independent.csv")
train = read_csv("data/data_set_ALL_AML_train.csv")
```

```python
from pandas import DataFrame


def remove_non_numeric(df: DataFrame) -> DataFrame:
    copy = df.copy()
    for col in copy.columns.copy():
        if copy[col].dtype == "object":
            del copy[col]
    return copy
```

```python
independent = remove_non_numeric(independent)
train = remove_non_numeric(train)
```
