```python
from pymongo import MongoClient

# Connect to the MongoDB server (default localhost:27017)
client = MongoClient("mongodb://localhost:27017/")

# Create or switch to the database "sample_db"
db = client.sample_db
```

```python
# List all databases
print(client.list_database_names())
```

```python
# Create collection users_profile (MongoDB creates collections automatically on insert, but to mimic explicitly:)
if "users_profile" not in db.list_collection_names():
    db.create_collection("users_profile")
```

```python
# Insert a single document
db.users_profile.insert_one({
    "userId": 1,
    "firstName": "John",
    "lastName": "Richard",
    "age": 26,
    "email": "john2992@mail.com"
})
```

```python
# Insert multiple documents
db.users_profile.insert_many([
    {
        "userId": 1,
        "firstName": "John",
        "lastName": "Richard",
        "age": 26,
        "email": "john2992@mail.com"
    },
    {
        "userId": 2,
        "firstName": "Kedar",
        "lastName": "Sans",
        "age": 29,
        "email": "kedar.sans@mail.com"
    },
    {
        "userId": 3,
        "firstName": "Chan",
        "lastName": "Kuli",
        "age": 27,
        "email": "chan.k@mail.com"
    }
])
```

```python
# Get all documents
list(db.users_profile.find({}))
```

```python
# Find documents where firstName is John
list(db.users_profile.find({"firstName": "John"}))
```

```python
# Find documents where firstName is John or Kedar
list(db.users_profile.find({"firstName": {"$in": ["John", "Kedar"]}}))
```

```python
# Find documents where firstName is John and age < 30
list(db.users_profile.find({
    "firstName": "John",
    "age": {"$lt": 30}
}))
```

```python
# Find documents where firstName is John or age < 30
list(db.users_profile.find({
    "$or": [
        {"firstName": "John"},
        {"age": {"$lt": 30}}
    ]
}))
```

```python
# Find documents where firstName is John and (age < 30 or lastName starts with 's')
import re
list(db.users_profile.find({
    "firstName": "John",
    "$or": [
        {"age": {"$lt": 30}},
        {"lastName": re.compile("^s", re.IGNORECASE)}
    ]
}))
```

```python
# Update one document (userId 1), set age to 20
db.users_profile.update_one({"userId": 1}, {"$set": {"age": 20}})
```

```python
# Update many documents where age is 20, set age to 35
db.users_profile.update_many({"age": 20}, {"$set": {"age": 35}})
```

```python
# Replace one document (userId 1)
db.users_profile.replace_one(
    {"userId": 1},
    {
        "userId": 1.0,
        "firstName": "Sam",
        "lastName": "Billings",
        "age": 26.0,
        "email": "sam2992@mail.com"
    }
)
```

```python
# Delete one document with userId 1
db.users_profile.delete_one({"userId": 1})
```

```python
# Delete all documents in the collection
db.users_profile.delete_many({})
```
